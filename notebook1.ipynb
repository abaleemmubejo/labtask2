```python
hours = float(input("Enter hours worked: "))
rate = float(input("Enter hourly rate: "))

if hours <= 40:
    total_pay = hours * rate
else:
    regular_pay = 40 * rate
    overtime_hours = hours - 40
    overtime_pay = overtime_hours * (rate * 1.5)
    total_pay = regular_pay + overtime_pay

print(f"Total Pay: ${total_pay:.2f}")
```

```python
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

marks = []
has_failed_subject = False

for i in range(1, 6):
    score = float(input(f"Enter marks for Subject {i} (out of 100): "))
    marks.append(score)
    if score < 40:
        has_failed_subject = True

total_marks = sum(marks)
percentage = (total_marks / 500) * 100

if percentage >= 80:
    grade = "A+"
elif percentage >= 70:
    grade = "A"
elif percentage >= 60:
    grade = "B"
elif percentage >= 50:
    grade = "C"
elif percentage >= 40:
    grade = "D"
else:
    grade = "F"

if has_failed_subject or grade == "F":
    result = "FAIL"
else:
    result = "PASS"

print("\n" + "=" * 40)
print(f"{'STUDENT MARKSHEET':^40}")
print("=" * 40)
print(f"Name: {name}")
print(f"Roll No: {roll_no}")
print("-" * 40)
for i, score in enumerate(marks, start=1):
    print(f"Subject {i}: {score:.2f} / 100")
print("-" * 40)
print(f"Total Marks: {total_marks:.2f} / 500")
print(f"Percentage : {percentage:.2f}%")
print(f"Grade      : {grade}")
print(f"Result     : {result}")
print("=" * 40)
```
